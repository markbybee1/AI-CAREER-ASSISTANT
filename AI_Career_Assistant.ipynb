#AI Career Assistant
Created by Mark Bybee
A Python-based career recommendation program that evaluates a user's interests and AI-related skills, recommends a potential career path, and calculates an AI readiness score.
##Features
- Collects user information
- Evaluates Python and AI experience
- Recommends AI-related career paths
- Calculates an AI readiness score
- Uses conditional logic to personalize results

```python
print ("Welcome to Mark's AI Career Assistant!")
name= input ("What's your name? ")
print ("Nice to meet you,", name)


career = input( "What career are you interested in? ")
experience = input ("Are you a beginner, intermediate, or advanced? ")
if experience.lower() == "beginner":
    print("Start learning the fundamnetals of", career)
elif experience.lower() == "intermediate":
      print("Start building projects related to", career)
elif experience.lower() == "advanced":
      print("Start applying for professional", career, "positions.")
else:
    print("Keep learning and building your skills!")

print("\n--- Skill Check ---")
python_skill  = input("Do you know any Pythom? yes/no: ")
ai_skill = input("Have you used AI tools before? yes/no: ")
if python_skill.lower() == "yes" and ai_skill.lower() == "yes":
    print("Great! You're ready to start building ai projects.")
elif python_skill.lower() =="yes":
    print("You know some Python. Next, start learning how AI models work.")
elif ai_skill.lower() == "yes":
    print("You have AI experience. Your next step is learning Python.")
else:
    print("Start with basic Python, then move into AI.")
print('\n--- AI Career Recommendation ---')
coding = input("Do you enjoy coding? yes/no: ")
data = input("Do you enjoy working with data? yes/no: ")
if coding.lower() == "yes" and data.lower() == "yes":
      print("Recommended path: AI / Machine Learning Developer")
elif data.lower() == "yes":
      print("Recommended path: Data Analyst")
elif coding.lower() == "yes":
      print("Recommended path: Python / AI Application Developer")
else:
      print("Recommended path: AI support or AI Operations")
print("\n--- Ai Readiness Scoree ---")
score = 0
if python_skill.lower() == "yes":
    score = score + 25
if ai_skill.lower() == "yes":
    score = score + 25
if coding.lower() == "yes":
    score = score + 25
if data.lower() == "yes":
    score = score + 25
print("Your Ai Readiness Score is:", score, "out of 100")
if score >= 80:
  print("Excellent start! You're ready to build more advanced Ai projects.")
elif score >=50:
  print("Good start! Keep developing your Python and AI skills.")
else:
    print("You're at the beginning of your AI journey. Keep learning!")
```
